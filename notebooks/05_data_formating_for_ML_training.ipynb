# Putting Data in correct format for Machine Learning Model Training

Importing required libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import (
    StandardScaler, OneHotEncoder, OrdinalEncoder, LabelEncoder
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
import pickle
```

Importing data

```python
df_telco = pd.read_csv("../data/feature_engineered_data.csv")
df_telco.head()
```

```python
df_telco.info()
```

## Going through the int64 and object columns to visualize values in order to adopt good way to encode the columns

```python
# Storing the concerned columns in a variable
obj_int_col = [col for col in df_telco.select_dtypes(include=["int64", "object"]).columns]

obj_int_col
```

```python
# Looping through the columns and printing unique values
for col in obj_int_col:
    print(f"Unique values for {col}:{df_telco[col].unique()}", end="\n\n")
```

## Drop the customerID column since it has no predictive value

```python
df_telco.drop(columns=["customerID"], axis=1, inplace=True)
```

```python
df_telco.info()
```

## Transforming target variable to binary 1/O

```python
df_telco["Churn"] = (df_telco["Churn"] == "Yes").astype("int")
```

```python
# Checking
df_telco["Churn"].unique(), df_telco["Churn"].dtype
```

## Transforming the gender column to contain binary data 1/0

```python
# transforming
df_telco["gender"] = (df_telco["gender"] == "Male").astype("int")
```

```python
# Checking
df_telco["gender"].unique(), df_telco["gender"].dtype
```

```python
df_telco.info()
```

## Defining list of pass through columns in the features

- These are columns in features which are already binary or numerical so do not need transformations for the moment

```python
passthrough_cols = [
    "gender", "SeniorCitizen", "Partner", "Dependents", "PhoneService", "OnlineSecurity", "OnlineBackup",
    "DeviceProtection", "TechSupport", "StreamingTV", "StreamingMovies", "PaperlessBilling", "IsMonthly", "IsSingle", "TotalServices",
    "tenure", "MonthlyCharges", "TotalCharges", "AvgMonthlyCharge", "ChargeDiff"
]
```

## Defining list of ordinal columns

```python
# list of columns
ordinal_cols = ["Contract", "TenureGroups"]

# list of categories
ordinal_categories = [
    ["Monthly", "One year", "Two year"],
    ["0-1", "1-2", "2-4", "4-6"]
]
```

## Defining list of nominal columns (no order in categories)

```python
ohe_cols = ["MultipleLines", "InternetService", "PaymentMethod"]
```

## List of numerical features to scale

```python
scale_cols = [
    "tenure", "MonthlyCharges", "TotalCharges", "AvgMonthlyCharge", "ChargeDiff", "TotalServices"
]
```

## Splitting the data into training and testing sets while maintaining same proportion of churn in both sets

Constructing the feature matrix

```python
# Feature matrix
X = df_telco.drop(columns=["Churn"])

# The target vector
y = df_telco["Churn"]
```

```python
X.columns
```

Actually splitting the data

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    random_state=42,
    stratify=y # To preserve churn ratios in both sets
)
```

```python
print(f"Train size : {X_train.shape[0]:,} rows")
print(f"Test size : {X_test.shape[0]:,} rows")
```

```python
print(f"Churn rate - Train : {y_train.mean()} | Test : {y_test.mean()}")
```

## Building the preprocessing pipeline

```python
preprocessor = ColumnTransformer(
    transformers=[
        # Scaling continuous numeric variables
        ("scaler",
         StandardScaler(),
         scale_cols),

         # Ordinal encode on ordered categorical features
         ("ordinal",
          OrdinalEncoder(categories=ordinal_categories),
          ordinal_cols),

        # One-hot encode nominal categoricals
        ("ohe",
         OneHotEncoder(drop="first", sparse_output=False, handle_unknown="ignore"),
         ohe_cols),

         # Passing binary columns as they are
         ("passthrough",
          "passthrough",
          [col for col in passthrough_cols if col not in scale_cols])
    ],
    verbose_feature_names_out=False
)
```

Fitting the preprocessor to the training data

```python
preprocessor.fit(X_train)
```

```python
feature_names = preprocessor.get_feature_names_out()
```

## Transforming the training and testing data

```python
# Training data
X_train_processed = preprocessor.transform(X_train)

# Testing data
X_test_processed = preprocessor.transform(X_test)
```

## Constructing and displaying final feature matrices infos

```python
# train
X_train_df = pd.DataFrame(X_train_processed, columns=feature_names)

# test
X_test_df = pd.DataFrame(X_test_processed, columns=feature_names)
```

# Export

```python
# Training and testing data
X_train_df.to_csv("../data/X_train.csv", index=False)
X_test_df.to_csv("../data/X_test.csv", index=False)
y_train.to_csv("../data/y_train.csv", index=False)
y_test.to_csv("../data/y_test.csv", index=False)
```
